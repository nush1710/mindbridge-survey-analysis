# MindBridge: Student Mental Health Survey Analysis

## Project Overview

MindBridge is an academic research project exploring the potential of an AI-powered mental health support chatbot designed for Indian college students.

This project analyses survey responses from 32 college students to understand their mental health support preferences, attitudes towards AI chatbots, adoption barriers and desired chatbot features.

## Objectives

- Understand the frequency of college-related stress among respondents.
- Explore students' comfort levels with AI-based mental health support.
- Identify concerns associated with AI chatbots.
- Analyse preferred chatbot features.
- Translate survey findings into potential product considerations.

## Tools & Technologies

- Python
- Pandas
- Matplotlib
- Google Colab

## Dataset

The analysis is based on 32 survey responses across 8 questions. Individual-level responses are excluded from the published project to protect respondent privacy.

## 1. Dataset Overview

- **Total responses:** 32
- **Total survey questions analysed:** 8
- **Data format:** CSV
- **Data type:** Categorical and text-based responses

The survey covered the following areas:

1. Frequency of college-related stress
2. Help-seeking behaviour
3. Barriers to seeking help
4. Comfort with AI mental health chatbots
5. Reasons for preferring AI over human counsellors
6. Concerns about AI chatbots
7. Perceived usefulness of an AI chatbot
8. Desired chatbot features

## 2. Data Cleaning

Data cleaning was performed on the original survey dataset using Python and Pandas. The following steps were carried out before analysing the aggregated results.

1. Loaded the CSV dataset into a Pandas DataFrame.
2. Removed unnecessary columns containing email addresses and timestamps.
3. Removed extra spaces from column names and text responses.
4. Replaced empty strings with missing-value markers.
5. Checked for missing values and verified the final dataset dimensions.

**Final dataset:** 32 rows × 8 columns.

No missing values were found after cleaning.

## 3. Exploratory Data Analysis (EDA)

Exploratory Data Analysis was performed to understand patterns in the survey responses.

The responses were analysed using frequency counts, percentages and bar charts.

### 3.1 Frequency of College-Related Stress

This analysis examines how frequently respondents experience college-related stress.

**Key Findings:**
- 37.5% experience stress sometimes.
- 37.5% experience stress very often.
- 21.88% experience stress rarely.
- 3.12% never experience stress.

**Key Insight:** 75% of respondents experience college-related stress sometimes or very often, highlighting the importance of accessible stress-management support.

```python
import pandas as pd
import matplotlib.pyplot as plt

stress_counts = pd.Series({
    "Sometimes": 12,
    "Very often": 12,
    "Rarely": 7,
    "Never": 1
})

stress_percentages = (stress_counts / 32) * 100

print("Response Counts:")
print(stress_counts)

print("\nResponse Percentages:")
print(stress_percentages.round(2))

stress_counts.plot(
    kind="bar",
    title="Frequency of College-Related Stress",
    xlabel="Stress Frequency",
    ylabel="Number of Students"
)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### 3.2 Student Comfort with AI Mental Health Chatbots

This analysis explores how comfortable respondents would feel discussing their mental health with an AI chatbot.

**Key Findings:**
- 50% are comfortable talking to an AI chatbot.
- 28.12% are unsure.
- 21.88% are not comfortable.

**Key Insight:** While half of the respondents are comfortable with AI-based support, the remaining respondents indicate a need to build trust and address concerns around using AI for mental health support.

```python
import pandas as pd
import matplotlib.pyplot as plt

ai_counts = pd.Series({
    "Yes": 16,
    "Not sure": 9,
    "No": 7
})

ai_percentages = (ai_counts / 32) * 100

print("Response Counts:")
print(ai_counts)

print("\nResponse Percentages:")
print(ai_percentages.round(2))

ai_counts.plot(
    kind="bar",
    title="Student Comfort with AI Mental Health Chatbots",
    xlabel="Responses",
    ylabel="Number of Students"
)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### 3.3 Concerns About AI Mental Health Chatbots

This analysis identifies the major concerns respondents have about using AI chatbots for mental health support.

**Key Findings:**
- 65.62% are concerned about receiving incorrect advice.
- 62.5% are concerned about the lack of human understanding.
- 59.38% have concerns about privacy and data security.
- 37.5% are concerned about over-dependence on AI.
- 9.38% selected "No concerns".

**Key Insight:** Accuracy, human understanding and data privacy are major areas that need attention when designing an AI mental health support product.

```python
import pandas as pd
import matplotlib.pyplot as plt

concern_counts = pd.Series({
    "Incorrect advice": 21,
    "Lack of human understanding": 20,
    "Privacy/data security": 19,
    "Over-dependence": 12,
    "No concerns": 3
})

concern_percentages = (concern_counts / 32) * 100

print("Response Counts:")
print(concern_counts)

print("\nResponse Percentages:")
print(concern_percentages.round(2))

concern_counts.plot(
    kind="bar",
    title="Concerns About AI Mental Health Chatbots",
    xlabel="Concerns",
    ylabel="Number of Students"
)

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## 4. Key Findings & Product Insights

Based on the analysis of 32 survey responses, the following insights emerged:

### 1. Need for Accessible Mental Health Support
75% of respondents experience college-related stress sometimes or very often, while 65.62% have not considered seeking help.

**Product Opportunity:** Explore accessible, low-pressure ways for students to engage with mental health support.

### 2. Trust Is Essential for AI Adoption
65.62% are concerned about incorrect advice, 62.5% about the lack of human understanding and 59.38% about privacy and data security.

**Product Opportunity:** Prioritise reliable responses, clear privacy communication and access to human support.

### 3. Focus on Practical Features
Stress relief exercises (59.38%) and emergency support (56.25%) received the highest feature preferences.

**Product Opportunity:** Consider these as potential feature priorities, alongside mood tracking and daily check-ins.

### 4. AI Should Complement Human Support
Although 50% are comfortable with AI chatbots, 25% explicitly said they would not prefer AI over human counsellors.

**Product Opportunity:** Position the chatbot as an additional support tool, not a replacement for professional counselling.

## 5. Limitations of the Study

- **Small Sample Size:** The analysis is based on only 32 responses and may not represent all college students.
- **Self-Reported Data:** Responses reflect participants' personal opinions and experiences.
- **Limited Generalisability:** Findings are exploratory and should not be generalised to the wider student population.
- **Multi-Select Questions:** Some questions allowed multiple answers, so percentages may not add up to 100%.
- **Feature Response Quality:** Some respondents entered names or unrelated text in the feature-preference question. Only four predefined feature categories were included in the analysis.
